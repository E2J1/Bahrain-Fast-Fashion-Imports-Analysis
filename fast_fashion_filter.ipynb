```python
import pandas as pd
cleaned_df = pd.read_csv('/content/bahrain_imports_cleaned.csv', dtype={'commodity_no': str}, engine='python')
```

```python
cleaned_df.info()
```

```python
# Get the most frequent commodity number
most_frequent = cleaned_df['commodity_no'].value_counts().idxmax()
frequency_count = cleaned_df['commodity_no'].value_counts().max()

print(f"The most frequent commodity is: {most_frequent}")
print(f"It appears {frequency_count} times in the dataset.")

# To see the top 5 most frequent commodities:
print("\nTop 5 Commodities:")
print(cleaned_df['commodity_no'].value_counts().head(5))
```

```python
commodity_61091000_df = cleaned_df[cleaned_df['commodity_no'] == '61091000']
display(commodity_61091000_df.head())
```

```python
commodity_61_df = cleaned_df[cleaned_df['commodity_no'].str.startswith('61')]
num_rows_61 = len(commodity_61_df)
print(f"The total number of rows for commodities starting with '61' is: {num_rows_61}")
```

```python
fast_fashion_codes = [
    '6103', '6104', '6105', '6106', '6109', '6110',
    '6203', '6204', '6205', '6206', '6208', '6207'
]

# Create a regex pattern to match commodity numbers starting with any of the codes
pattern = '|'.join([f'^{code}' for code in fast_fashion_codes])

# Filter the DataFrame
fast_fashion_df = cleaned_df[cleaned_df['commodity_no'].str.contains(pattern, regex=True)]

# Get the total number of rows
total_fast_fashion_rows = len(fast_fashion_df)

print(f"The total number of rows for the specified fast-fashion apparel commodities is: {total_fast_fashion_rows}")
```

```python
display(fast_fashion_df.head())
```

```python
unique_fast_fashion_commodity_descriptions = fast_fashion_df['commodity'].unique().tolist()
print("Unique fast-fashion commodity descriptions:")
for description in unique_fast_fashion_commodity_descriptions:
    print(f"- {description}")
print(f"\nTotal unique fast-fashion commodity descriptions: {len(unique_fast_fashion_commodity_descriptions)}")
```

```python
fast_fashion_df.to_csv('fast_fashion_imports.csv', index=False)
print("Fast fashion data exported successfully to 'fast_fashion_imports.csv'.")
```
